```python
#merging the datasets because we have two datasets for one movie
#id and title are same
movies=movies.merge(credits,on='title')
movies.head(3)
```

```python
#separating columns for recommendation 
#genres
#id
#keywords
#title
#popularity
#overview
#cast 
#crew
movies=movies[['id','genres', 'keywords','title','popularity','overview','cast','crew']]
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum() #no duoplicate values
```

```python
movies.dropna(subset=['overview'], inplace=True) #dropping 3 null values of the overview column
```

```python
movies[('genres')].describe()
```

```python
import ast

movies['genres'] = movies['genres'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies['keywords'] = movies['keywords'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies['cast'] = movies['cast'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies[['genres', 'keywords','cast']].head(1)
```

```python
movies['cast'] = movies['cast'].apply(lambda x: x[:3])
```

```python
movies['cast'].head()
```

```python
movies['keywords'].apply(len).head()
```

```python
movies['crew'] = movies['crew'].apply(
    lambda x: [x] if x else []
)
```

```python
movies['crew'].head(10)
```

```python
movies.head()
```

```python
#transformation for removing space in between same words
movies['genres'] = movies['genres'].apply( lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply( lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply( lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply( lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['tags'] = ( movies['overview'].apply(lambda x: [x]) + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew'])
```

```python
new_df = movies[['id','title','tags']]
```

```python
new_df['tags']=new_df['tags'].apply(lambda x:" ".join(x)) #changed tags list into string
```
