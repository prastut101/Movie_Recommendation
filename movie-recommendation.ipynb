```python
import numpy as np
import pandas as pd

credits = pd.read_csv('/Users/prastutupreti/AI_prastut/datasets/tmdb_5000_credits.csv')
movies = pd.read_csv('/Users/prastutupreti/AI_prastut/datasets/tmdb_5000_movies.csv')

movies.head(3)
```

```python
# merging the datasets because we have two datasets for one movie
# id and title are same

movies = movies.merge(credits, on='title')

movies.head(3)
```

```python
# separating columns for recommendation
# genres
# id
# keywords
# title
# popularity
# overview
# cast
# crew

movies = movies[['id', 'genres', 'keywords', 'title', 'popularity', 'overview', 'cast', 'crew']]

movies.info()
```

```python
movies.dropna(subset=['overview'], inplace=True)
```

```python
import ast
```

```python
movies['genres'] = movies['genres'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies['keywords'] = movies['keywords'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies['cast'] = movies['cast'].apply(
    lambda x: [i['name'] for i in ast.literal_eval(x)]
)
```

```python
movies['cast'] = movies['cast'].apply(lambda x: x[:3])
```

```python
movies['cast'].head()
```

```python
movies['keywords'].apply(len).head()
```

```python
def get_director(x):
    for i in ast.literal_eval(x):
        if i['job'] == 'Director':
            return [i['name']]
    return []
```

```python
movies['crew'] = movies['crew'].apply(get_director)
```

```python
movies['crew'].head(10)
```

```python
movies['genres'] = movies['genres'].apply(
    lambda x: [i.replace(" ", "_") for i in x]
)

movies['keywords'] = movies['keywords'].apply(
    lambda x: [i.replace(" ", "_") for i in x]
)

movies['cast'] = movies['cast'].apply(
    lambda x: [i.replace(" ", "_") for i in x]
)

movies['crew'] = movies['crew'].apply(
    lambda x: [i.replace(" ", "_") for i in x]
)
```

```python
movies['crew'].head()
```

```python
movies['tags'] = (
    movies['overview'].apply(lambda x: [x])
    + movies['genres']
    + movies['keywords']
    + movies['cast']
    + movies['crew']
)
```

```python
movies.head()
```

```python
new_df = movies[['id', 'title', 'tags']].copy()
```

```python
new_df['tags'] = new_df['tags'].apply(
    lambda x: " ".join(x)
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem_tags(tags):
    return [ps.stem(word) for word in tags.split()]
```

```python
new_df['tags'] = new_df['tags'].apply(stem_tags)
```

```python
from sklearn. metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
similarity[0][1]
```

```python
def recommend(movie):

    # Remove unnecessary spaces
    movie = movie.strip()

    # Find the movie
    matches = movies[movies['title'].str.lower() == movie.lower()]

    # Check if movie exists
    if matches.empty:
        return []

    # Get movie index
    movie_index = matches.index[0]

    # Get similarity scores
    distances = similarity[movie_index]

    # Get top 5 similar movies
    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    recommended_movies = []

    for i in movies_list:
        recommended_movies.append(movies.iloc[i[0]].title)

    return recommended_movies
```

```python
recommend('Batman Begins')
```
